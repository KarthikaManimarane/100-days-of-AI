````python
from langchain_core.tools import tool
from langchain_google_genai import ChatGoogleGenerativeAI
from dotenv import load_dotenv
import os

print("Imports ready!")
````

````python
load_dotenv("../../WEEK 1/DAY 1/.env")

print("API key loaded:", os.getenv("GEMINI_API_KEY") is not None)
````

````python
@tool
def add(a: int, b: int) -> int:
    """Add two numbers."""
    return a + b

print("Tool:", add.name)
print("Description:", add.description)
````

````python
llm = ChatGoogleGenerativeAI(
    model="gemini-3.6-flash",
    google_api_key=os.getenv("GEMINI_API_KEY")
)

print("Gemini ready!")
````

````python
from langchain.agents import create_agent

agent = create_agent(
    model=llm,
    tools=[add],
    system_prompt="You are a helpful calculator assistant. Use the add tool when needed."
)

print("Agent created!")
````

````python
result = agent.invoke({
    "messages": [
        {"role": "user", "content": "What is 25 + 75?"}
    ]
})

print(result["messages"][-1].content)
````

````python
for msg in result["messages"]:
    print(type(msg).__name__)
    print(msg.content)
    print("-" * 40)
````
