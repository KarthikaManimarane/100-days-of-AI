````python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict

print("LangGraph ready!")
````

````python
class State(TypedDict):
    number: int
    result: str

print("State defined!")
````

````python
def large_number(state: State):
    return {"result": "Large number"}

def small_number(state: State):
    return {"result": "Small number"}

print("Branch nodes created!")
````

````python
def decide(state: State):
    if state["number"] > 5:
        return "large"
    else:
        return "small"

print("Decision function created!")
````

````python
graph = StateGraph(State)

graph.add_node("large_number", large_number)
graph.add_node("small_number", small_number)

graph.add_conditional_edges(
    START,
    decide,
    {
        "large": "large_number",
        "small": "small_number"
    }
)

graph.add_edge("large_number", END)
graph.add_edge("small_number", END)

app = graph.compile()

print("Conditional graph compiled!")
````

````python
result = app.invoke({
    "number": 8,
    "result": ""
})

print(result)
````

````python
result = app.invoke({
    "number": 3,
    "result": ""
})

print(result)
````

## Conclusion

In this project, I learned how conditional edges work in LangGraph.

I created a simple State Graph where a Python routing function checks the value of `number` and decides which node should execute.

If the number is greater than 5, the graph goes to the `large_number` node. Otherwise, it goes to the `small_number` node.

### Key Learning

- Normal edges follow a fixed path.
- Conditional edges allow a graph to branch.
- A routing function can decide which path to take.
- `add_conditional_edges()` is used to create conditional routing.
- The routing decision in this example is handled by Python.
- Multiple branches can be created from a single decision.
