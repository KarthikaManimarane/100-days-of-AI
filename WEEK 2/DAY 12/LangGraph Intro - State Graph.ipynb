````python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict

print("LangGraph ready!")
````

````python
class State(TypedDict):
    message: str

print("State defined!")
````

````python
def node1(state: State):
    return {"message": "Hello"}

print("Node 1 created!")
````

````python
def node2(state: State):
    return {"message": state["message"] + " World"}

print("Node 2 created!")
````

````python
graph = StateGraph(State)

graph.add_node("node1", node1)
graph.add_node("node2", node2)

graph.add_edge(START, "node1")
graph.add_edge("node1", "node2")
graph.add_edge("node2", END)

app = graph.compile()

print("Graph compiled!")
````

````python
result = app.invoke({"message": ""})

print(result)
````

## Conclusion

In this project, I learned the basics of LangGraph and how a State Graph works.

I created a simple two-node graph where the state is passed from one node to another. Node 1 updates the message to "Hello", and Node 2 receives that state and adds " World".

### Key Learning

- State stores shared information used by the graph.
- Nodes are functions that perform work and update the state.
- Edges define the flow between nodes.
- `START` represents the beginning of the graph.
- `END` represents the end of the graph.
- `StateGraph` is used to build a state-based workflow.
- `compile()` prepares the graph to be executed.
