````python
from langchain_core.tools import tool
from langchain_google_genai import ChatGoogleGenerativeAI
from dotenv import load_dotenv
import os

print("Imports ready!")
````

````python
load_dotenv("../../WEEK 1/DAY 1/.env")

llm = ChatGoogleGenerativeAI(
    model="gemini-3.6-flash",
    google_api_key=os.getenv("GEMINI_API_KEY")
)

print("Gemini ready:", os.getenv("GEMINI_API_KEY") is not None)
````

````python
@tool
def add(a: int, b: int) -> int:
    """Add two numbers."""
    return a + b

print(add.name)
print(add.description)
````

````python
llm_with_tools = llm.bind_tools([add])

print("Tool bound!")
````

````python
response = llm_with_tools.invoke("What is 25 + 75?")

print(response.tool_calls)
````

````python
tool_call = response.tool_calls[0]
tool_result = add.invoke(tool_call)

print("Tool result:", tool_result.content)
````

````python
from langchain_core.messages import HumanMessage

messages = [
    HumanMessage(content="What is 25 + 75?"),
    response,
    tool_result
]

final_response = llm_with_tools.invoke(messages)

print(final_response.text)
````
