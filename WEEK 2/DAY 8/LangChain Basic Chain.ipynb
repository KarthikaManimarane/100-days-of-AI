```python
from langchain_core.prompts import PromptTemplate

prompt = PromptTemplate.from_template(
    "Summarize this text in 2-3 sentences:\n\n{text}"
)

print("Prompt template created!")
```

```python
chain = prompt | llm

print("Chain created!")
```

```python
text = """
Python is a high-level programming language known for its simple syntax.
It is widely used in web development, automation, data science,
machine learning, and software development.
"""

result = chain.invoke({"text": text})

print(result.content)
```

```python
print(result.content[0]["text"])
```

```python
user_text = input("Enter text to summarize: ")

result = chain.invoke({"text": user_text})

print("\nSummary:")
print(result.content[0]["text"])
```

```python
try:
    user_text = input("Enter text to summarize: ")

    if not user_text.strip():
        print("Please enter some text.")
    else:
        result = chain.invoke({"text": user_text})

        print("\nSummary:")
        print(result.content[0]["text"])

except Exception as e:
    print("Something went wrong:", e)
```

## Conclusion

In this project, I built a basic text summarizer using LangChain.

I learned how to create a reusable PromptTemplate, connect it to a Gemini LLM, and combine them into a chain using the pipe operator.

The chain accepts user input through `invoke()`, sends the formatted prompt to Gemini, and returns a summary.

### Key Learning

- PromptTemplate creates reusable prompts with placeholders.
- The LLM generates the response.
- The `|` operator connects components in a LangChain pipeline.
- `invoke()` runs the chain with input values.
- The same chain can be reused with different user inputs.
