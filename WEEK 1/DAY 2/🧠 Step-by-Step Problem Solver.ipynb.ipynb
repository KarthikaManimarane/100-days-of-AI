```python
from dotenv import load_dotenv
from google import genai
import os

load_dotenv("../DAY 1/.env")

client = genai.Client(api_key=os.getenv("GEMINI_API_KEY"))

print("Gemini connected!")
```

```python
problem = """
A shop gives a 20% discount on a ₹2,500 jacket.
After the discount, 18% GST is added to the discounted price.
What is the final price?
"""
```

```python
normal_prompt = f"""
Solve this problem and give me the final answer.

{problem}
"""
```

```python
normal_response = client.models.generate_content(
    model="gemini-3.6-flash",
    contents=normal_prompt
)

print(normal_response.text)
```

```python
step_prompt = f"""
Solve this problem step by step.
Show each calculation clearly before giving the final answer.

{problem}
"""
```

```python
step_response = client.models.generate_content(
    model="gemini-3.6-flash",
    contents=step_prompt
)

print(step_response.text)
```

```python
%%writefile README.md
# Day 02 - Chain-of-Thought Prompting

## What I learned
- Chain-of-Thought prompting asks the model to solve a problem step by step.
- It can make the model's response more detailed and structured.
- It does not train the model or guarantee a correct answer.

## What I built
A Step-by-Step Problem Solver using Gemini. I gave Gemini the same multi-step pricing problem using a normal prompt and a step-by-step prompt, then compared the responses.

## What was tricky
Understanding that step-by-step prompting changes how the model approaches the current request; it does not permanently train the model.
```
