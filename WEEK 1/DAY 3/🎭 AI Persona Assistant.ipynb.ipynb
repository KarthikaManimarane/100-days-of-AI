```python
from dotenv import load_dotenv
from google import genai
import os

load_dotenv("../DAY 1/.env")

client = genai.Client(api_key=os.getenv("GEMINI_API_KEY"))

print("Gemini connected!")
```

```python
user_prompt = """
Explain why learning Python is useful for someone starting a career in AI.
"""
```

```python
teacher_system = """
You are a friendly and patient AI teacher.
Explain concepts in simple language.
Use a small example when helpful.
Avoid unnecessary technical jargon.
"""
```

```python
mentor_system = """
You are a strict technical AI mentor.
Explain concepts precisely and technically.
Focus on practical skills, tools, and career relevance.
Avoid oversimplifying important technical details.
"""
```

```python
friend_system = """
You are a flirty friend.
Explain things in a relaxed, conversational way.
Use simple language and relatable examples.
You can use light humor when appropriate.
"""
```

```python
teacher_response = client.models.generate_content(
    model="gemini-3.6-flash",
    contents=[
        {
            "role": "user",
            "parts": [
                {"text": teacher_system + "\n\n" + user_prompt}
            ]
        }
    ]
)

print(teacher_response.text)
```

```python
mentor_response = client.models.generate_content(
    model="gemini-3.6-flash",
    contents=mentor_system + "\n\n" + user_prompt
)

print(mentor_response.text)
```

```python
friend_response = client.models.generate_content(
    model="gemini-3.6-flash",
    contents=friend_system + "\n\n" + user_prompt
)

print(friend_response.text)
```
