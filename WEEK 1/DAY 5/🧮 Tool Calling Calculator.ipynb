```python
from dotenv import load_dotenv
from google import genai
import os

load_dotenv("../DAY 1/.env")

client = genai.Client(api_key=os.getenv("GEMINI_API_KEY"))

print("Gemini connected!")
```

```python
def calculator(a: float, b: float, operation: str) -> float:
    if operation == "add":
        return a + b
    elif operation == "subtract":
        return a - b
    elif operation == "multiply":
        return a * b
    elif operation == "divide":
        return a / b
    else:
        raise ValueError("Invalid operation")
```

```python
print(calculator(10, 5, "add"))
print(calculator(10, 5, "subtract"))
print(calculator(10, 5, "multiply"))
print(calculator(10, 5, "divide"))
```

```python
tools = [calculator]
```

```python
user_prompt = "What is 25 multiplied by 12?"
```

```python
response = client.models.generate_content(
    model="gemini-3.6-flash",
    contents=user_prompt,
    config={
        "tools": tools
    }
)

print(response)
```

```python
print(response.candidates[0].content.parts)
```

```python
tool_result = calculator(25, 12, "multiply")

print(tool_result)
```

```python
final_response = client.models.generate_content(
    model="gemini-3.6-flash",
    contents=f"""
The calculator tool returned this result:

{tool_result}

Give the user a clear final answer to their original question.
"""
)

print(final_response.text)
```

## Conclusion

In this project, I learned how tool/function calling works with an LLM.

Gemini was given access to a Python calculator function. When the user asked a calculation question, Gemini requested the calculator tool with the required arguments.

The Python application then executed the function and produced the result. That result was passed back to Gemini, which generated the final response.

### Key Learning

- Gemini decides when a tool is needed.
- Gemini provides the tool name and required arguments.
- The Python application executes the actual function.
- The tool result is returned to Gemini.
- Gemini uses the result to produce the final response.
- Tool calling allows LLM applications to interact with external functions and data.
