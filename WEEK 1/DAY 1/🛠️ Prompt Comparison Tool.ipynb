```python
from google import genai
import os

client = genai.Client(api_key=os.getenv("GEMINI_API_KEY"))

response = client.models.generate_content(
    model="gemini-3.6-flash",
    contents="Say hello in one sentence."
)

print(response.text)
```

```python
review = "The food was amazing, but the service was slow."

zero_shot_prompt = f"""
Classify the sentiment of this review as Positive, Negative, or Neutral.

Review: {review}
"""
```

```python
zero_shot_response = client.models.generate_content(
    model="gemini-3.6-flash",
    contents=zero_shot_prompt
)

print(zero_shot_response.text)
```

```python
few_shot_prompt = f"""
Classify the sentiment of each review as Positive, Negative, or Neutral.

Examples:

Review: "The food was delicious and the staff were friendly."
Sentiment: Positive

Review: "The food was terrible and the staff were rude."
Sentiment: Negative

Review: "The food was good but the service was slow."
Sentiment: Neutral

Now classify this review:

Review: {review}
Sentiment:
"""
```

```python
few_shot_response = client.models.generate_content(
    model="gemini-3.6-flash",
    contents=few_shot_prompt
)

print(few_shot_response.text)
```

```python
%%writefile README.md
# Day 01 - Zero-Shot vs Few-Shot Prompting

## What I learned
- Zero-shot prompting gives the model a task without examples.
- Few-shot prompting gives the model a task along with a few examples.
- Few-shot prompting can guide the model toward the desired format or behavior.

## What I built
A Prompt Comparison Tool using Gemini that sends the same sentiment-classification task using both zero-shot and few-shot prompting and compares the results.

## What was tricky
Understanding that few-shot examples are provided in the prompt for context; they do not permanently train the model.
```
