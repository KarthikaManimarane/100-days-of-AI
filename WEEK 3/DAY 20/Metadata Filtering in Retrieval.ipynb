````python
documents = [
    "YOLOv10 is used for vehicle damage detection.",
    "The system classifies damage as minor, moderate, or severe.",
    "The model achieved a mAP of 88.7%.",
    "The chatbot helps users understand vehicle damage reports."
]

metadata = [
    {"section": "technology"},
    {"section": "severity"},
    {"section": "results"},
    {"section": "chatbot"}
]

for document, meta in zip(documents, metadata):
    print(document)
    print(meta)
    print()
````

````python
import chromadb

client = chromadb.Client()

collection = client.get_or_create_collection(
    name="metadata_demo"
)

print("Collection ready!")
````

````python
collection.add(
    ids=[f"doc_{i}" for i in range(len(documents))],
    documents=documents,
    metadatas=metadata
)

print("Documents added:", collection.count())
````

````python
query = "What are the damage levels?"

results = collection.query(
    query_texts=[query],
    n_results=2,
    where={"section": "severity"}
)

print("Retrieved documents:\n")

for i, doc in enumerate(results["documents"][0]):
    print(f"--- Document {i + 1} ---")
    print(doc)
    print("Metadata:", results["metadatas"][0][i])
    print()
````

````python
query = "What was the model performance?"

results = collection.query(
    query_texts=[query],
    n_results=2,
    where={"section": "severity"}
)

print("Retrieved documents:")
for doc in results["documents"][0]:
    print(doc)
````

````python
query = "What was the model performance?"

results = collection.query(
    query_texts=[query],
    n_results=2
)

print("Without metadata filter:\n")

for doc in results["documents"][0]:
    print(doc)
    print()
````

# Conclusion

Today I learned how metadata filtering can be used to control which documents are considered during vector retrieval.

I added metadata tags such as `technology`, `severity`, `results`, and `chatbot` to documents stored in Chroma. I then used the `where` parameter to filter retrieval based on metadata.

For example, using `where={"section": "severity"}` restricted the search to documents tagged as `severity`, even when the question was related to another section.

### Retrieval Flow

Question → Vector Search + Metadata Filter → Relevant Documents

### Key Takeaways

- Metadata provides additional information about stored documents.
- Chroma supports filtering vector search using metadata.
- The `where` parameter can restrict which documents are searched.
- Metadata filtering can improve retrieval control and precision.
- Semantic similarity and metadata filtering can be combined.
