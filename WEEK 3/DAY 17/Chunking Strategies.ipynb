```python
from langchain_community.document_loaders import PyPDFLoader

print("PDF loader ready!")
```

```python
pdf_path = r"C:\Users\PC\Downloads\karthika journal.pdf"

loader = PyPDFLoader(pdf_path)

pages = loader.load()

print("Number of pages:", len(pages))
```

```python
print(pages[0].page_content[:1500])
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

splitter = RecursiveCharacterTextSplitter(
    chunk_size=200,
    chunk_overlap=20
)

chunks_200 = splitter.split_documents(pages)

print("Number of chunks:", len(chunks_200))
```

```python
for i, chunk in enumerate(chunks_200[:3]):
    print(f"\n--- Chunk {i + 1} ---")
    print(chunk.page_content)
```

```python
splitter_500 = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50
)

chunks_500 = splitter_500.split_documents(pages)

print("Number of chunks:", len(chunks_500))
```

```python
splitter_1000 = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=100
)

chunks_1000 = splitter_1000.split_documents(pages)

print("Number of chunks:", len(chunks_1000))
```

```python
print("===== 200 CHARACTER CHUNK =====")
print(chunks_200[0].page_content)

print("\n===== 500 CHARACTER CHUNK =====")
print(chunks_500[0].page_content)

print("\n===== 1000 CHARACTER CHUNK =====")
print(chunks_1000[0].page_content)
```
