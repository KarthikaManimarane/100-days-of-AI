```python
import os
from dotenv import load_dotenv
from langchain_google_genai import GoogleGenerativeAIEmbeddings

load_dotenv(r"C:\Users\PC\Desktop\100 DAYS OF AI\WEEK 1\DAY 1\.env")

print("Setup complete!")
```

```python
embeddings = GoogleGenerativeAIEmbeddings(
    model="models/gemini-embedding-001",
    google_api_key=os.getenv("GEMINI_API_KEY")
)

print("Embedding model ready!")
```

```python
sentences = [
    "I love learning artificial intelligence.",
    "I enjoy studying AI and machine learning.",
    "Python is my favorite programming language.",
    "I like eating pizza.",
    "The weather is beautiful today."
]

print(sentences)
```

```python
vectors = embeddings.embed_documents(sentences)

print("Number of embeddings:", len(vectors))
print("Dimensions:", len(vectors[0]))
```

```python
from sklearn.metrics.pairwise import cosine_similarity

similarity = cosine_similarity(vectors)

print(similarity)
```

```python
for i in range(len(sentences)):
    for j in range(i + 1, len(sentences)):
        print(f"{i+1} ↔ {j+1}: {similarity[i][j]:.4f}")
```

# Conclusion

In this project, I learned the basics of text embeddings.

Embeddings convert text into numerical vectors that represent the meaning of the text.

I generated embeddings for 5 sentences and used cosine similarity to compare how semantically similar the sentences are.

## Key Learning

- Embeddings represent text as numerical vectors.
- Similar meanings generally produce more similar vectors.
- Cosine similarity can be used to compare embeddings.
- A similarity score closer to 1 means the vectors are more similar.
- Comparing a sentence with itself gives a similarity of 1.
