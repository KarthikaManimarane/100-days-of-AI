# Conclusion

Today I learned how to use Chroma as a local vector database.

I generated 3072-dimensional embeddings using Gemini and stored five sentences along with their embeddings in a Chroma collection. I then converted a query into an embedding and used Chroma to retrieve the most semantically similar documents.

I also learned that the embedding dimensions must match between stored vectors and query vectors. When Chroma was using 384-dimensional embeddings while Gemini produced 3072-dimensional embeddings, the query failed. I fixed this by explicitly storing the Gemini embeddings in Chroma.

### Key Takeaways

- Chroma can store embeddings locally.
- Embeddings allow text to be compared based on meaning.
- Vector databases can retrieve semantically similar information.
- Stored and query embeddings must have matching dimensions.
- Lower vector distance generally means greater similarity.
