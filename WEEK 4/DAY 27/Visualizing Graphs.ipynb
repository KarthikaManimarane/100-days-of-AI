````python
import networkx as nx
import matplotlib.pyplot as plt
````

````python
G = nx.DiGraph()

G.add_edge("Alice", "Google", relation="works_at")
G.add_edge("Google", "Gemini", relation="develops")
G.add_edge("Alice", "Python", relation="uses")
````

````python
pos = nx.spring_layout(G, seed=42)
````

````python
nx.draw(G, pos, with_labels=True)
plt.show()
````

````python
edge_labels = nx.get_edge_attributes(G, "relation")
print(edge_labels)
````

````python
nx.draw(G, pos, with_labels=True)
nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels)
plt.show()
````

````python
from pyvis.network import Network
````

````python
net = Network(notebook=True, directed=True)

net.from_nx(G)
````

````python
net.set_options("""
var options = {
  "edges": {
    "arrows": {
      "to": {
        "enabled": true
      }
    }
  }
}
""")
````

````python
net.show("knowledge_graph.html")
````

````python
for edge in net.edges:
    source = edge["from"]
    target = edge["to"]
    edge["label"] = G[source][target]["relation"]
````

````python
net.show("knowledge_graph.html")
````

## Matplotlib vs PyVis

| Feature | Matplotlib | PyVis |
|---|---|---|
| Visualization | Static | Interactive |
| Move nodes | No | Yes |
| Zoom | Limited | Yes |
| Hover interaction | No | Yes |
| Relationship labels | Yes | Yes |
| Best for | Analysis & learning | Exploring knowledge graphs |

Matplotlib is useful for creating simple and static graph visualizations.

PyVis is better when we want to explore a knowledge graph interactively.

````python
net = Network(notebook=True, directed=True, height="600px", width="100%")

net.from_nx(G)

for edge in net.edges:
    source = edge["from"]
    target = edge["to"]
    edge["label"] = G[source][target]["relation"]

net.show("knowledge_graph.html")
````
