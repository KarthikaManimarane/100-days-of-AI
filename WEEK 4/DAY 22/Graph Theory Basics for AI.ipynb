```python
social = nx.Graph()
```

```python
social.add_nodes_from(["Alice", "Bob", "Charlie", "David", "Emma"])
```

```python
social.add_edges_from([
    ("Alice", "Bob"),
    ("Alice", "Charlie"),
    ("Bob", "Charlie"),
    ("Bob", "David"),
    ("Charlie", "Emma")
])
```

```python
print(social.edges)
```

```python
for person in social.nodes:
    print(person, "→", social.degree(person))
```

```python
nx.draw(social, with_labels=True)
plt.show()
```

```python
path = nx.shortest_path(social, "Alice", "Emma")
print(path)
```
