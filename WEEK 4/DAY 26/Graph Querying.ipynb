````python
def find_relationship(subject, relation):
    results = []

    for source, target, data in G.edges(data=True):
        if source == subject and data["relation"] == relation:
            results.append(target)

    return results
````

````python
answer = find_relationship("Alice", "uses")

print(answer)
````

````python
answer = find_relationship("Google", "develops")

print("Google develops:", answer)
````
