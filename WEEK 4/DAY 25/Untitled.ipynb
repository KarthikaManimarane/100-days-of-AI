```python
triples = [
    ("Alice", "works_at", "Google"),
    ("Google", "develops", "Gemini"),
    ("Alice", "uses", "Python")
]

print(triples)
```

```python
for subject, relation, object_ in triples:
    G.add_edge(subject, object_, relation=relation)
```

```python
print("Nodes:", list(G.nodes))
print("Edges:", list(G.edges(data=True)))
```

```python
import matplotlib.pyplot as plt

nx.draw(G, with_labels=True)
plt.show()
```

```python
edge_labels = nx.get_edge_attributes(G, "relation")

pos = nx.spring_layout(G)

nx.draw(G, pos, with_labels=True)
nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels)

plt.show()
```

```python
text = """
Alice works at Google.
Google develops Gemini.
Alice uses Python.
"""
```

```python
triples = [
    ("Alice", "works_at", "Google"),
    ("Google", "develops", "Gemini"),
    ("Alice", "uses", "Python")
]
```

```python
G = nx.DiGraph()

for subject, relation, object_ in triples:
    G.add_edge(subject, object_, relation=relation)
```

```python
print("Nodes:", list(G.nodes))
```

```python
print("Relationships:")

for subject, object_, data in G.edges(data=True):
    print(subject, "→", data["relation"], "→", object_)
```

```python
print("Alice's connections:")

for neighbor in G.successors("Alice"):
    print(neighbor)
```

```python
print("Number of nodes:", G.number_of_nodes())
```

```python
print("Number of edges:", G.number_of_edges())
```

```python
print("Knowledge Graph:")

for subject, object_, data in G.edges(data=True):
    print(f"{subject} --[{data['relation']}]--> {object_}")
```

# Conclusion

Today I learned how to convert extracted triples into a NetworkX graph.

I learned that a triple consists of a subject, relation, and object. The subject and object become nodes in the graph, while the relation is stored as an edge between them.

For the mini-project, I converted multiple triples into a directed NetworkX graph and analyzed its nodes and relationships.

### Key Takeaways

- Triples follow the Subject → Relation → Object structure.
- Subjects and objects become graph nodes.
- Relations become edges between nodes.
- Directed graphs preserve the direction of relationships.
- NetworkX can be used to build, visualize, and analyze knowledge graphs.
- Converting text into triples is an important foundation for Knowledge Graphs and Graph RAG.
